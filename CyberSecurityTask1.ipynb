```python
def caesar_encrypt(text, shift):
    """
    Encrypt text using Caesar cipher with given shift value.

    Args:
        text (str): The text to encrypt
        shift (int): The number of positions to shift each letter

    Returns:
        str: The encrypted text
    """
    result = ""

    for char in text:
        if char.isalpha():
            # Determine if uppercase or lowercase
            ascii_offset = ord('A') if char.isupper() else ord('a')
            # Shift the character and wrap around using modulo
            shifted_char = chr((ord(char) - ascii_offset + shift) % 26 + ascii_offset)
            result += shifted_char
        else:
            # Keep non-alphabetic characters unchanged
            result += char

    return result

def caesar_decrypt(text, shift):
    """
    Decrypt text using Caesar cipher with given shift value.

    Args:
        text (str): The text to decrypt
        shift (int): The number of positions the text was shifted during encryption

    Returns:
        str: The decrypted text
    """
    # Decryption is just encryption with negative shift
    return caesar_encrypt(text, -shift)

def get_valid_shift():
    """
    Get a valid shift value from user input.

    Returns:
        int: Valid shift value
    """
    while True:
        try:
            shift = int(input("Enter shift value (integer): "))
            return shift
        except ValueError:
            print("Please enter a valid integer.")

def display_menu():
    """Display the main menu options."""
    print("\n" + "="*40)
    print("         CAESAR CIPHER TOOL")
    print("="*40)
    print("1. Encrypt text")
    print("2. Decrypt text")
    print("3. Exit")
    print("="*40)

def main():
    """Main program loop."""
    print("Welcome to the Caesar Cipher Tool!")

    while True:
        display_menu()

        choice = input("Choose an option (1-3): ").strip()

        if choice == '1':
            # Encryption
            print("\n--- ENCRYPTION ---")
            message = input("Enter the text to encrypt: ")
            shift = get_valid_shift()

            encrypted_text = caesar_encrypt(message, shift)

            print(f"\nOriginal text:  {message}")
            print(f"Shift value:    {shift}")
            print(f"Encrypted text: {encrypted_text}")

        elif choice == '2':
            # Decryption
            print("\n--- DECRYPTION ---")
            message = input("Enter the text to decrypt: ")
            shift = get_valid_shift()

            decrypted_text = caesar_decrypt(message, shift)

            print(f"\nEncrypted text: {message}")
            print(f"Shift value:    {shift}")
            print(f"Decrypted text: {decrypted_text}")

        elif choice == '3':
            print("\nThank you for using the Caesar Cipher Tool!")
            break

        else:
            print("\nInvalid choice. Please select 1, 2, or 3.")

        # Ask if user wants to continue
        if choice in ['1', '2']:
            continue_choice = input("\nWould you like to perform another operation? (y/n): ").lower().strip()
            if continue_choice not in ['y', 'yes']:
                print("\nThank you for using the Caesar Cipher Tool!")
                break

def demo():
    """
    Demonstration function showing how the Caesar cipher works.
    """
    print("\n--- CAESAR CIPHER DEMONSTRATION ---")

    # Example encryption
    original = "Hello, World!"
    shift_val = 3
    encrypted = caesar_encrypt(original, shift_val)
    decrypted = caesar_decrypt(encrypted, shift_val)

    print(f"Original message: {original}")
    print(f"Shift value: {shift_val}")
    print(f"Encrypted: {encrypted}")
    print(f"Decrypted back: {decrypted}")
    print(f"Match original: {original == decrypted}")

if __name__ == "__main__":
    # Uncomment the line below to see a demonstration
    # demo()

    # Run the main program
    main()
```
